Question 2

```python
print("Checking for missing values in the dataset:")
display(mis_datos.isnull().sum())
```

The code tells us that: Engine size, cylinders, horsepower, MPG_City, MPG_ Highway, weight, wheelbase, and length, contain missing values and it gives us the amount of missing values it contains as well.

```python
print("Replace missing values with the mean of the respective columns:")
for column in mis_datos.columns:
    if mis_datos[column].isnull().sum() > 0:
        mean_value = mis_datos[column].mean()
        mis_datos[column].fillna(mean_value, inplace=True)
display(mis_datos.isnull().sum())
        
```

Here we can see that the previous columns that had missing values now appear to have none.

Question 3

```python
print("Which car manufacturer has the largest number of cars in the dataset? Which car manufacturer has the smallest number of cars in the dataset?")
manufacturer_counts = mis_datos['Make'].value_counts()
largest_manufacturer = manufacturer_counts.idxmax()
smallest_manufacturer = manufacturer_counts.idxmin()
print(f"The car manufacturer with the largest number of cars is: {largest_manufacturer} with {manufacturer_counts.max()} cars.")
print(f"The car manufacturer with the smallest number of cars is: {smallest_manufacturer} with {manufacturer_counts.min()} cars.")
```

Here we can see that the manufacturer that produces the largest number of vehicles is Toyota with 28 cars and the smallest number is Hummer with 1 car

Question 4

```python
print("How many cars are produced in each region with front-wheel drive, rear-wheel drive, and all-wheel drive?")
drive_counts = (
    mis_datos.groupby(["Origin", "DriveTrain"])
    .size()
    .unstack(fill_value=0))
sum_counts = drive_counts.sum(axis=1)
drive_counts["Total"] = sum_counts

display(drive_counts)
```

Here we can see the total of cars produced in each region based on all drives. In Asia the total is 158, in Europe 123, and in USA 147.

```python
print("Highest suggested retail price:")
highest_price = mis_datos["MSRP"].max()
display(highest_price)

print("Lowest suggested retail price:")
lowest_price = mis_datos["MSRP"].min()
display(lowest_price)

print("Average city miles per gallon:")
average_city_mpg = mis_datos["MPG_City"].mean()
display(average_city_mpg)

print("Average highway miles per gallon (MPG):")
average_highway_mpg = mis_datos["MPG_Highway"].mean()
display(average_highway_mpg)
```
